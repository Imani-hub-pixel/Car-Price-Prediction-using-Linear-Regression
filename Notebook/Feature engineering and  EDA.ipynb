## EDA

Univariate analysis

```python
sns.countplot(car_df['fueltype'])
plt.show()
```

```python
sns.catplot(x='aspiration',data=car_df,kind='count')
plt.show()
```

```python
sns.catplot(x='drivewheel',data=car_df,kind='count')
plt.show()
```

```python
sns.boxplot(y='wheelbase',data=car_df)
plt.show()
```

```python
sns.boxplot(y='curbweight',data=car_df)
plt.show()
```

Bivariate analysis

```python
car_df.info()
```

```python
sns.relplot(data=car_df,x='horsepower',y="price",kind="scatter",col="fueltype")
plt.show()
```

```python
sns.relplot(data=car_df,x="enginesize",y="price",kind="scatter")
plt.show()
```

Multivariate analysis

```python
## Density plot to understand the skweness of the numerical data.
sns.set_theme()
car_df.hist(bins=10,figsize=(20,15))
plt.show()
```

```python
sns.pairplot(car_df)
plt.show()
```
